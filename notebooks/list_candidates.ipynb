```python
from pathlib import Path
from bs4 import BeautifulSoup
```

```python
list_files = Path("../data/raw/official_data/voting_paris_candidates_2020").glob(
    "*.html"
)
cand_dict = {}
for f in list_files:
    htmlfile = open(f, "r", encoding="utf-8")
    data = htmlfile.read()
    soup = BeautifulSoup(data, "html.parser")
    all_res = soup.find_all("td")
    for i in range(0, len(all_res), 3):
        cand_dict[all_res[i + 2].get_text()] = all_res[i + 1].get_text()
```
